## 6. Yeni kapsamlı ölçüm (isteğe bağlı)
Teslim kanıtı zaten var; yeniden çalıştırmak gerekmiyor. Yeni ölçüm ayrı klasöre yazılır; kaynak/model kilidi ve her tur checkpoint korunur.

```python
RUN_NEW_EVALUATION=False
if RUN_NEW_EVALUATION:
    from alfiq_b.evaluate import run_evaluation
    EVAL=WORK/'runs'/'new-live-acceptance-v4-8'
    EVAL.mkdir(parents=True,exist_ok=True)
    hashes={p.relative_to(PROJECT).as_posix():hashlib.sha256(p.read_bytes()).hexdigest()
            for folder in ['alfiq_b','schemas','data'] for p in (PROJECT/folder).rglob('*')
            if p.is_file() and p.suffix in {'.py','.json'}}
    lock={'source_sha256':hashes,'configuration':live_configuration('ollama')}
    lock_path=EVAL/'source_lock.json'
    if lock_path.exists():assert json.loads(lock_path.read_text(encoding='utf-8'))==lock
    else:c.atomic_json_write(lock_path,lock)
    report=run_evaluation(EVAL,mode='live',provider='ollama',allow_live=True,repetitions=5,
                          resume=(EVAL/'progress.json').exists(),llm_options={'num_ctx':4096,'timeout':120})
    print(json.dumps(report,ensure_ascii=False,indent=2))
else:print('Yeni canlı eval kapalı; kayıtlı kabul raporu bölüm 3 içinde.')
```

## 7. Yeni kayıtları arşivle
Yeni demo/ölçüm JSON'larını yerel ZIP'e koyar. Kaynak/eval teslim ZIP'i ayrı dosyadır.

```python
from IPython.display import FileLink,display
destination=WORK/'ALFIQ_v4_8_new_evidence.zip'
with zipfile.ZipFile(destination,'w',zipfile.ZIP_DEFLATED) as archive:
    for folder in [WORK/'runs',WORK/'runtime'/'reports']:
        if folder.exists():
            for path in folder.rglob('*'):
                if path.is_file() and path.suffix in {'.json','.jsonl','.txt'}:
                    archive.write(path,path.relative_to(WORK).as_posix())
print('Arşiv:',destination.resolve())
display(FileLink(str(destination)))
```
